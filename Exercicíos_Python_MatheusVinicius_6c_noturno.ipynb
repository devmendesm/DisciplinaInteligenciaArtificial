```python
# 1. Hello, World!
# Utilizei a função nativa do python
print("Hello, World!")
```

```python
# 2. Soma de Dois Números
# Recebe a entrada do usuario com input() ja convertendo para o tipo int
# Crio uma variavel soma e faço a impressão
n1 = int(input("Digite um número: "))
n2 = int(input("Digite mais um número: "))
soma = n1 + n2

print(soma)
```

```python
# 3. Cálculo da Área do Círculo
# Recebe a entrada do usuario e converte para o tipo float
# Utiliza o valor de pi (perguntar se é o valor exato) e utiliza (**) que representa potencia
raio = float(input("Digite o raio do círculo: "))
area = 3.14 * raio ** 2

print(f"Area do círculo: {area}")
```

```python
# 4. Conversão de Temperatura
# Recebe a entrada do usuario e converte para o tipo float e utiliza a fórmula para conversão
tempCelsius = float(input("Digite a temperatura em Celsius: "))
tempFah = tempCelsius * 9 / 5 + 32

print(f"{tempFah}°F")
```

```python
# 5. Tabuada
# Recebe a entrada do usuario convertendo para inteiro
# Faz um laço de repetição for com um limitador de inicio e fim, acaba em 11 porque o range ignora o ultimo número
n = int(input("Digite um número inteiro: "))

for x in range(1, 11):
  print(f"{n} x {x} = {n * x}")
```

```python
# 6. Números Pares
# Faz uma verificação dentro do laço para saber se o resto da divisao (%) por 2 é 0,
# se for igual, é par e exibe, se não for, não exibe
for x in range(1, 21):
  if (x % 2 == 0):
    print(x)
```

```python
# 7. Número Ímpar ou Par
# Recebe a entrada do usuario convertendo para int
# Faz uma condição se o resto da divisao for 0 imprime par do contrário imprime ímpar
numero = int(input("Digite um número inteiro: "))

if (numero % 2 == 0):
  print(f"O número {numero} é par.")
else:
  print(f"O número {numero} é ímpar.")
```

```python
# 8. Verificação de Palíndromo
# Utiliza o recurso de fatiamento do python, [::-1] percorre a palavra de tras pra frente com -1
palavra = input("Digite uma palavra: ")
invertida = palavra[::-1]

if (palavra == invertida):
  print(f"A palavra {palavra} é um palíndromo")
else:
  print(f"A palavra {palavra} não é um palíndromo")
```

```python
# 9. Fatorial de um Número
# A variavel fatorial serve para ir acumulando o valor a ser multiplicado
# dentro do range o (numero +1) é necessário para que a conta não pare em um numero antes do digitado
numero = int(input("Digite um número inteiro: "))
fatorial = 1

for x in range(1, numero + 1):
  fatorial = fatorial * x

print(f"O fatorial de {numero} é: {fatorial}")
```

```python
# 10.Sequência de Fibonacci
# Define os 2 primeiros numeros da sequencia e dentro do laço,
# faz a soma e reatribui os valores das variaveis A e B,
# onde A passa a ser B e B passa a ser o próximo numero da sequência
a = 0
b = 1

for x in range(10):
  print(a)

  proximo = a + b
  a = b
  b = proximo
```
