```python
# 21.Busca Binária ****** DUVIDA
# A busca começa pelo meio da lista, por isso utiliza-se inicio meio e fim
procurado = int(input("Digite um número inteiro: "))
numeros = [2, 5, 8, 12, 15, 20, 30]
inicio = 0
fim = len(numeros) - 1

while inicio <= fim:
  # // retorna a divisao em int e nao float
  meio = (inicio + fim) // 2

  # se o procurado for maior ele vai eliminar a esquerda e procurar na direita
  # se for menor ele vai eliminar a direita e procurar somenta na esquerda
  if numeros[meio] == procurado:
    print(f"Número encontrado na posição {meio}")
    break
  elif procurado > numeros[meio]:
    inicio = meio + 1
  else:
    fim = meio - 1
```

```python
# 22. Criptografia de Cifra de César ******** DUVIDA
# A função ord() transforma um caracter em um número A = 65, a chr() transforma o numero em letra
# A função isalpha() retorna true/false se houver ou nao numeros dentro da string
def criptaMensagem(texto, deslocamento):
  resultado = ""

  for letra in texto:
    if (letra.isalpha()):
      codigo = ord(letra) - ord('A')
      codigo = (codigo + deslocamento) % 26
      resultado += chr(codigo + ord('A'))

    else:
      resultado += letra
  return resultado

mensagem = input("Digite a mensagem: ").upper()

print(criptaMensagem(mensagem, 3))
```

```python
# 23. Descriptografar Cifra de César  ********** DUVIDA
# aqui segue a lógica igualmente, porém devemos tirar o deslocamento do codigo ao inves de acrescentar
def criptaMensagem(texto, deslocamento):
  resultado = ""

  for letra in texto:
    if (letra.isalpha()):
      codigo = ord(letra) - ord('A')
      codigo = (codigo - deslocamento) % 26
      resultado += chr(codigo + ord('A'))

    else:
      resultado += letra
  return resultado

mensagem = input("Digite a mensagem: ").upper()

print(criptaMensagem(mensagem, 3))
```

```python
# 30.Jogo da Forca
# importa a lib random para utilizar a função choice que vai sortear a palavra, e começa com 5 tentativas,
# só sai do laço while quando atingir 0
import random
tentativas = 5
palavras = ["PYTHON", "JAVA", "JAVASCRIPT", "HTML"]
palavra = random.choice(palavras)
letrasDescobertas = []

while (tentativas != 0):
  for letra in palavra:

    # Mostra as letras ou _ de acordo com a lista de descobertas, se acertar adiciona na lista, se errar -1 tentativa
    if letra in letrasDescobertas:
      print(letra, end=" ")
    else:
      print("_", end=" ")

  print(f"\nTentativas restantes: {tentativas}")
  tentativa = input("Digite uma letra: ").upper()

  if tentativa in palavra:
    letrasDescobertas.append(tentativa)
    print(letrasDescobertas)
  else:
    tentativas -= 1
    print(letrasDescobertas)

  # Verifica se todas (all) as letras dentro da lista descobertas pertencem a palavra, se sim ganhou e para o loop
  if all(letra in letrasDescobertas for letra in palavra):
    print("Parabens! Você venceu")
    break

if (tentativas == 0):
  print(f"Você perdeu, a palavra era {palavra}")

```

```python
# 32.Jogo da Velha
# Função que imprime o tabuleiro, o join() junta os elementos de uma string com algum separador escolhido
def mostrarTabuleiro(tabuleiro):
    print()
    for linha in tabuleiro:
        print(" | ".join(linha))


def verificarVitoria(tabuleiro):
    # verifica as linhas do tabuleiro
    for linha in tabuleiro:
        if linha[0] == linha[1] == linha[2] != " ":
            return True

    # verifica as colunas
    for i in range(3):
        if tabuleiro[0][i] == tabuleiro[1][i] == tabuleiro[2][i] != " ":
            return True

    # verifica a diagonal da esquerda para direita
    if tabuleiro[0][0] == tabuleiro[1][1] == tabuleiro[2][2] != " ":
        return True

    # verifica a diagonal da direita para esquerda
    if tabuleiro[0][2] == tabuleiro[1][1] == tabuleiro[2][0] != " ":
        return True

    return False


def verificarEmpate(tabuleiro):

    for linha in tabuleiro:
        if " " in linha:
            return False

    return True


def main():

    tabuleiro = [
        [" ", " ", " "],
        [" ", " ", " "],
        [" ", " ", " "]
    ]

    jogador = "X"

    while True:

        mostrarTabuleiro(tabuleiro)

        print(f"Jogador {jogador}")
        linha = int(input("Linha (0-2): "))
        coluna = int(input("Coluna (0-2): "))

        if tabuleiro[linha][coluna] != " ":
            print("Posição ocupada!")
            continue

        tabuleiro[linha][coluna] = jogador

        if verificarVitoria(tabuleiro):
            mostrarTabuleiro(tabuleiro)
            print(f"Jogador {jogador} venceu!")
            break

        if verificarEmpate(tabuleiro):
            mostrarTabuleiro(tabuleiro)
            print("Empate!")
            break

        # Faz a troca da vez do jogador
        if jogador == "X":
            jogador = "O"
        else:
            jogador = "X"


main()
```

```python
# 40. Lista de Compras
# uma lista para armazenar os itens adicionados pelo usario e um dicionario para produtos e valor
lista = []
itens = {
    "Arroz": 12, "Feijão": 15,
    "Café": 30, "Carne": 90
    }

# Aqui o programa continuará em funcionamento até o usuario digitar "4"
while True:
  # Menu para usuario ter noção das opções que ele pode usar
  print("\n1 - Adicionar")
  print("2 - Remover")
  print("3 - Listar")
  print("4 - Sair")

  opcao = input("Digite uma opção: ")

  if (opcao == "1"):
    print()
    for item, preco in itens.items():
      print(item, "-", preco)

    item = input("Digite o item: ")
    lista.append(item)

  elif (opcao == "2"):
    print()
    item = input("Digite o item para remover: ")

    if (item in lista):
      lista.remove(item)
    else:
      print("item não encontrado")

  elif (opcao == "3"):
    print()
    print("\nLista de compras:")

    for item in lista:
      print(item)

  elif (opcao == "4"):
    print()
    print("Compras finalizadas")
    print(f"Sua lista de compras: {lista}")
    break

  else:
    print("Opção inválida")

```

```python
# 46. Calcular Média e Desvio Padrão ********** DUVIDA
# Faz a soma dos numeros da lista e divide pelo valor de quantos numeros tem dentro dela
import math

numeros = [2, 4, 6, 8, 10]
media = sum(numeros) / len(numeros)
soma = 0

# calculo para achar a distancia de cada numero da média, elevar ao quadrado somando tudo
# divide esse valor pela quantidade de numeros e faz a raiz quadrada
for numero in numeros:
  # a soma vai ser igual ao numero da lista - media elevado ao quadrado (**)
  soma += (numero - media) ** 2

desvio = math.sqrt(soma / len(numeros))

print(f"Média: {media}")
print(f"Desvio padrão: {desvio}")
```

```python
# 47. Contar Números em Lista
# percorre os numeros da lista e vai adicionando em um dicionario incrementando com 1 se ja estiver,
# e recebendo 1 se ainda nao estiver no dicionario
numeros = [1, 4, 5, 7, 10, 7, 15, 4, 4]
contador = {}

for numero in numeros:
  if numero in contador:
    contador[numero] += 1
  else:
    contador[numero] = 1

# exibe formatado onde numero recebe a chave do dicionario e quant o valor da chave
for numero, quant in contador.items():
  print(f"{numero}: {quant} vez(s)")
```

```python
# 48.Ordenação por Inserção ********* DUVIDA
# o Implemente o algoritmo de ordenação por inserção (Insertion Sort).
numeros = [5, 3, 4, 1, 2]

# percorre a lista, começa em 1, pois considera que o 0 ja esta certo
for i in range(1, len(numeros)):
  # numero que vamos arrumar
  chave = numeros[i]
  # numero anterior
  j = i - 1

  # enquanto existir elemento a esquerda e a chave for menor, continua movendo
  while j >= 0 and chave < numeros[j]:
    print(j)
    # passa o numero para a direita
    numeros[j + 1] = numeros[j]
    j -= 1

  # coloca a chave no lugar
  numeros[j + 1] = chave

print(numeros)
```

```python
# 49.Soma dos Dígitos
numero = int(input("Digite um número: "))
soma = 0

while numero > 0:
  # o resto da divisão por 10 de um numero sempre restara o ultimo digito
  ultimo = numero % 10
  soma += ultimo

  # a divisao do numero por 10 vai retornar um decimal e o "//" elimina tudo apos a virgula
  numero = numero // 10
  print(numero)

print(f"A soma é: {soma}")
```

```python
# 50. Cálculo de Frequência de Letras
# replace() para transformar os espaços, para nao entrarem na conta
# percorre cada letra do texto incluindo no dicionario e incrementando valor caso ja tenha
texto = input("Digite uma frase: ").upper().replace(" ", "")
frequencia = {}

for letra in texto:
  if letra in frequencia:
    frequencia[letra] += 1
  else:
    frequencia[letra] = 1

for letra, quant in frequencia.items():
  print(f"{letra}: {quant}")
```
